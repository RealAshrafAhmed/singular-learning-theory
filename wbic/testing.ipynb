```python
import models 
import utils
import wbic
import numpy as np

np.random.seed(42)
X = np.random.normal(loc=0, scale=1, size=250)
model = models.tempered_normal_mixture_model(
    X,
    n_components=1,
    beta = 1/np.log(X.size),
)
gmm = wbic.BayesianModel(
    model=model,
    observations=X,
)
wbics = []
for i in range(125):
    gmm.sample(
        draws=1000,
        tune=100,
        chains=4,
        #max_treedepth=50,
        #target_accept=.995
    )
    wbics.append(gmm.WBIC())
#print(f'estimated LC: {LC}')
```

```python
def WBIC(self):
    if self.inference_data is None:
       print("Warning: Posterior has not yet been sampled; sampling with default parameters now.")
       self.sample(
        draws=2500,
        tune=2500,
        chains=4,
        max_treedepth=50,
        target_accept=.995
       )
    # log_likelihood has shape (n_chain, n_draw, n_data)
    # so summing over axis=2 is computing log p(w|X) = sum(log p(w|xi))
    ll = np.sum(self.inference_data.log_likelihood['likelihood'].values, axis=2)
    n_chain, n_draw = ll.shape
    random_mask = np.random.randint(0, n_chain, size=n_draw)
    ll_samples = [ll[random_mask[i], i] for i in range(n_draw)]
    self.WBIC = np.mean(ll_samples)
    wbic_array = []
    for i in range(n_chain):
        wbic_i = np.mean(ll[i,:])
        wbic_array.append(wbic_i)
        print(f"WBIC of chain {i}: {wbic_i}")

    return self.WBIC, wbic_array

wbic, wbic_array = WBIC(gmm)
```

```python
def learning_coefficient(self, wbic):
    n_data = self.observations.size
    ## Takio & Suzuki
    if True:
        Tn = self.empirical_loss()
        #print(n_data*Tn)
        LC = (-wbic - n_data*Tn)/np.log(n_data)
        return LC

lcs = []
for wbic in wbics:
    lc = learning_coefficient(gmm, wbic)
    lcs.append(lc)

print(np.mean(lcs))
print(np.var(lcs))
```
